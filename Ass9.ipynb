```python
# Write a function that accepts a list of numbers and returns the largest and smallest number. 
def com(list1):
    large = max(list1)
    small = min(list1)
    return large, small
num = [34, 4, 23, 34, 98]
largest , smallest = com(num) 
print("smallest - ", smallest,"largest - ", largest)
```

```python
# Write a function using  *args to calculate the sum, average, maximum, and minimum of any number of integers.  
def cal(*args):
    total = sum(args)
    maximum = max(args)
    minimum = min(args)
    average = total / len(args)

    return total, maximum, minimum, average


total, maximum, minimum, average = cal(45, 3332, 56, 34, 63, 22, 95)

print("Sum     :", total)
print("Maximum :", maximum)
print("Minimum :", minimum)
print("Average :", average)
```

```python
# Write a function that accepts a string and returns the number of uppercase letters, lowercase letters, digits.  
def return_number(str1):
    uppercase = 0
    lowercase = 0
    digit = 0
    for i in str1:
        if i.isupper():
            uppercase += 1
        elif i.islower():
            lowercase += 1
        elif i.isdigit():
            digit += 1
    return lowercase, uppercase, digit


uppercase, lowercase, digit = return_number("GaganPathare8888")  
print("uppercase - ", uppercase, "lowercase - ", lowercase, "digits - ", digit )          
            
    
```

```python
#  Write a function that checks whether a given string is a Palindrome. 
def ispalindrome(str1):
    str2 = "".join(reversed(str1))
    
    if str1 == str2:
        print("its palindrome")
    else:
        print("its not palindrome")   
        
ispalindrome("tatya")
        
```

```python
# . Write a function that accepts a list of integers and returns a new list containing only even numbers.  
def only_even(list1):
    list2 = []
    for i in list1:
        if i % 2 == 0:
            list2.append(i)
    print("even numbers - ",list2)
num_list = [43, 56, 89, 23, 54, 22]
only_even(num_list)
```

```python
# 9. Write a function that accepts a dictionary of students and their marks, then returns the name of the topper and the highest marks. 
def topper_name(**kvalues):
    name = max(kvalues, key=kvalues.get)
    marks = kvalues[name]
    return name, marks

students = { "sarthak": 7, "omkar": 78, "gagan":999}
topper_name(**students)
    
        
        
        
    
```

```python
# 10. Write a function that accepts a sentence and returns the frequency of each word using a dictionary. 
def freq_counter(str1):
    
    freq = {}
    for i in str1:
        if i in freq:
            freq[i ] += 1
        else:
            freq[i] = 1
    return(freq)
freq_counter("dghkltresxcvbn434")
        
```

```python

# 11. Write a function that accepts a list of integers and returns: 
# • Sum of all numbers  • Count of even numbers  • Count of odd numbers  
def counter(numbers):
    total = sum(numbers)
    even = 0
    odd = 0
    for i in numbers:
        if i%2 == 0:
            even += 1
        else:
            odd += 1
    return total, even, odd


list1 = [56,131,54,555,13,662,123]
t, e, o = counter(list1)      
print("total = ", t)
print("even = ", e )      
print("odd = ", o )      
```
