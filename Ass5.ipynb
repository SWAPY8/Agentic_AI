```python
# 1. Write a program to count the number of even and odd numbers in a list.
listone = [89, 43, 77, 24, 90, 23]
even = 0
odd = 0

for i in listone:
    if i % 2 == 0:
        even += 1
    else:
        odd += 1    
print(even)  
print(odd)
```

```python
# 2. Write a program to create a new list containing only the positive numbers from a given list. 
# - - -  - - Input: [-10, 20, -5, 40, 0, 15] Output: [20, 40, 15] 

oldlist = [-10, 20, -5, 40, 0, 15]
newlist = []
for i in oldlist:
        if i > 0:
            newlist.append(i)
print(newlist)
```

```python
# 3. Write a program to remove all duplicate elements from a list while preserving the original order. 
# - - -  - - Input: [10,20,30,20,40,10,50] Output: [10,20,30,40,50] 

oldlist = [10,20,30,20,40,10,50]
newlist = []
for i in oldlist:
    if i not in newlist:
        newlist.append(i)
print(newlist)         
```

```python
# 4. Write a program to find the second largest element in a list. - - -  - - [12,45,67,89,54] Output: 67 
oldlist = [12,45,67,89,54]
largest = 0
seclargest = 0
for i in oldlist:
    if i > largest :
        seclargest = largest
        largest = i
    elif i != largest and i != seclargest:
        seclargest = i 
print(seclargest)           
```

```python
# 5. Write a program to find the second smallest element in a list. 
oldlist = [12,45,67,89,54]
smallest = max(oldlist)
secsmallest = max(oldlist)
for i in oldlist:
    if i < smallest :
        secsmallest = smallest
        smallest = i
    elif i != smallest and i < secsmallest:
        secsmallest = i 
print(secsmallest)           
```

```python
# 6. Write a program to count the frequency of every element in a list. 
# - Input: - -  - - - - [1,2,3,2,1,4,1] Output: 1 -> 3 2 -> 2 3 -> 1 

listone = [1,2,3,2,1,4,1]
visited =[]
for i in listone:
    if i not in visited:
        count = 0

        for j in listone:
            if j == i:
                count += 1
            
        print(i, " -> ", count)    
        visited.append(i)
```

```python
# 7. Write a program to reverse a list without using the reverse() method or slicing ([::-1]). 
sortedlist = [ "78", "34", "13", "89", "200"]
newlist = []

# sortedlist = sorted(oldlist)
for i in range(len(sortedlist) - 1, -1, -1):
    newlist.append(sortedlist[i])

print(newlist) 


```

```python
# 8. Write a program to check whether a list is a Palindrome. - Input: - -  - - [1,2,3,2,1] Output: Palindrome 
oldlist = [1,2,9,8,2,1]
newlist =[]

for i in range(len(oldlist)-1, -1, -1):
    newlist.append(i)

if oldlist != newlist:
    print( "Not a Palindrome" )  
else:
    print("palindrome")           
```

```python
# 9. Write a program to find all common elements between two lists.
#  - List1 = [10,20,30,40] - -  - - List2 = [20,40,60] Output: [20,40] 

list1 = [10,20,30,40]
List2 = [20,40,60]
common = []
for i in list1:
    for j in List2:
        if j == i:
            common.append(i)
print(common)
```

```python

# 10. Write a program to separate all even and odd numbers into two different lists. .
# - - -  - - - [10,15,20,25,30] Output: Even = [10,20,30] Odd = [15,25] 

oldlist = [10,15,20,25,30]
list_of_odd = []
list_of_even = []

for i in oldlist:
    if i % 2 == 0:
        list_of_even.append(i)
    else:
        list_of_odd.append(i)
print("even_list = ",list_of_even )     
print("list_of_odd = ", list_of_odd)       
```
