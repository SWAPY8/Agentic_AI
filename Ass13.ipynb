```python
# 6. Write a program to read a file and create another file containing only the lines that contain the word "Python".
with open("Student_details.txt", "r") as file7 , open("filtered_stud.txt", "w") as file8:
    for lines in file7:
        if "python" in lines:
            file8.write(lines)
            
        
```

```python
# 8. Write a program to read a file and find the longest word present in it. 

with open("sample.txt", "r") as file10:
    temp = file10.read ()  
var10 = temp.split()
largest = max(var10, key = len)
print(largest)
        
```
