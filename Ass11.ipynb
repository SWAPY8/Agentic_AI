```python
# 1. Write a program to handle ZeroDivisionError when dividing two numbers entered by the user. 
try:
    num1 = int(input("Enter number :")) 
    num2 = int(input("Enter number :"))
    res = num1 / num2
except ZeroDivisionError as err:
    
    print(err)
else:
    print(res)

```

```python
# 2. Write a program to handle ValueError when the user enters a non-numeric value for age.  


try:
    name = input("Enter number :") 
    age = int(input("Enter number :"))
    
except ValueError as err:
    
    print("enter numeric values ",err)
else:
    print(name, age)
```

```python
# 6. Write a program that demonstrates the use of try, except, else, and finally blocks.
try:
    student_name = input("enter ur name: ")  
    student_dept = input("enter ur dept: ")  
    student_age = int(input("enter ur age: ") ) 
    student_contact = int(input("enter ur contact: ")  )
except Exception as err:
    print(err)
else:
    print("student_name : ", student_name)
    print("student_dept : ", student_dept)
    print("student_age : ", student_age)
    print("student_contact : ", student_contact)
    print()
finally:
    print("student details")


```
