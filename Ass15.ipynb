```python
# 2. Product Class – __eq__() Create a Product class with name and price. Implement __eq__() to check whether two products have the same price. 

class Products:
    def __init__(self, name, price):
        self.name = name
        self.price = price
    
    def __eq__(self, other):
        return self.price == other.price

p1 = Products("BMW", 60000)
p2 = Products("pagani", 600000)
print("Is equal = ", p1 == p2)
```

```python
# 3. Shopping Cart – __len__() Create a ShoppingCart class containing a list of products. Implement __len__() so that len(cart) returns the number of products in the cart. 

class ShoppingCart:
    def __init__(self,cart):
        self.cart = cart
    def __len__(self):
        return len(self.cart)
list1 = ["fs","sdf","fsf","ger","fs","sdf","fsf"]
s1 = ShoppingCart(list1)
print(len(s1))
        
        
        
        
```

```python
# 4. Bank Account – __str__() Create a BankAccount class with account holder name, account number, and balance. Implement __str__() to display account information when the object is printed. 

class BankAccount:
    def __init__(self, name, acc_no, balance):
        self.name = name
        self.acc_no = acc_no
        self.balance = balance
    def __str__(self):
        return f"Account Holder Name = {self.name}, Account Number - {self.acc_no}, Balance = {self.balance}."
b1 = BankAccount("Swapnil", 47577478588, 99999999)
print(b1)
```

```python
# 5. Number Class – __add__() Create a Number class that stores one number. Implement __add__() so that two objects can be added 

class Number:
    def __init__(self, num1):
        self.num1 = num1
    def __add__(self, other):
        return self.num1 + other.num1
n1 = Number(73)
n2 = Number(45)
print("Output = ", n1 + n2)
```

```python

# 6. Create a Number class and implement: __add__()  __sub__()  __mul__()  __truediv__()  __mod__() \

class Numbers:
    def __init__(self, num1):
        self.num1 = num1
    
    def __add__(self, other):
        return self.num1 + other.num1
    def __sub__(self, other):
        return self.num1 - other.num1
    def __mul__(self, other):
        return self.num1 * other.num1
    def __truediv__(self, other):
        return self.num1 / other.num1
    def __mod__(self, other):
        return self.num1 % other.num1
n1 = Numbers(375)
n2 = Numbers(2)

print(f"Addition = ",n1+n2 )
print(f"Subtraction = ",n1-n2 )
print(f"Multiplication = ",n1*n2 )
print(f"Division = ",n1/n2 )
print(f"Modulus = ",n1%n2 )

```
