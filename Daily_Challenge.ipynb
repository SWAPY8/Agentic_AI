```python
# combine the list and find median
list1 = [31,62,89]
list2 = [46, 85, 96,32]
combined = list1 + list2

print(combined)

mid = len(combined) // 2
# print(mid)

if len(combined) % 2 != 0:
    print(combined[mid])
else :
    prev = combined[mid-1] 
    next = combined[mid]
    avg = (prev+next ) / 2 
    print("avg=", avg)

```

```python
# move_zeroes(num)

def move_zeroes(num):
    zeros = []
    output = []
    for i in num:
        if i == 0:
            zeros.append(0)
        else:
            output.append(i)
    output.sort()       
    output.extend(zeros)
    return output
    
nums = [243, 0, 98, 0, 345, 45, 0]

move_zeroes(nums)

```

```python
# input - aaabbbbccccc
# o/p  - cccccaaabbbb

input = "aaabccbbbccccc"
list1  = []
list2 = []

for i in range(len(input)-1, 0, -1):
    if input[i] == input[i-1]:
        list1.append(input[i])
    else:
        list1.append(input[i])
        break


for i in range(len(input)-len(list1)):
    list2.append(input[i])
    
list1.extend(list2)

output = "".join(list1)
# print(list1)
print("Output = ",output)
    
```
