```python
# move_zeroes(num)

def move_zeroes(num):
    zeros = []
    output = []
    for i in num:
        if i == 0:
            zeros.append(0)
        else:
            output.append(i)
    output.sort()       
    output.extend(zeros)
    return output
    
nums = [243, 0, 98, 0, 345, 45, 0]

move_zeroes(nums)

```

```python
# input - aaabbbbccccc
# o/p  - cccccaaabbbb

input = "aaabccbbbcccccffff"
list1  = []
list2 = []

for i in range(len(input)-1, 0, -1):
    if input[i] == input[i-1]:
        list1.append(input[i])
    else:
        list1.append(input[i])
        break


for i in range(len(input)-len(list1)):
    list2.append(input[i])
    
list1.extend(list2)

output = "".join(list1)
# print(list1)
print("Output = ",output)
    
```
