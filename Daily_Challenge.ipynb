```python
# combine the list and find median
list1 = [31,62,89]
list2 = [46, 85, 96]
combined = list1 + list2

print(combined)

mid = len(combined) // 2

if len(combined) % 2 != 0:
    print(combined[mid])
else :
    prev = combined[mid-1] 
    next = combined[mid]
    avg = (prev+next ) / 2 
    print("avg=", avg)

```
