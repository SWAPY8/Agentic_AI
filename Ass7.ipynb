```python
# 1. Write a program to find the union of two sets without using the union() method. 
set1 = {1, 2, 4, 5, 6, 7, 8, 9, 43}
set2 = {23, 4, 7, 5, 9, 42, 65, 56}
common = set()


for i in set1:
    if  i not in common:
        common.add(i)    

for j in set2:
    if  j not in common:
        common.add(j)    

print("Union = ", common) 
```

```python
# . Write a program to remove all duplicate elements from a list using a set. 
list1 = [ 2, 65, 4, 5, 6, 2, 8, 9, 43, 43, 23, 2]

print(set(list1))

```

```python
# 3. Write a program to find all common elements between two lists using sets.
list1 = [1, 2, 4, 5, 6, 7, 8, 9, 43]
list2 = [23, 4, 7, 5, 9, 42, 65, 56]

temp = set(list1).intersection(set(list2))
print(temp)
```

```python
# 4. Write a program to count the number of unique elements in a list using a set.

list1 = [ 2, 65, 4, 5, 6, 2, 8, 9, 43, 43, 23, 2]
temp = set(list1)
print("count - ", len(temp))
```

```python
# 5. Write a program to find the missing numbers between 1 to N using sets.
# - Input:- N = 10 - Numbers = {1,2,3,5,7,8,10} Output:- {4,6,9}
N = 10
Numbers = {1,2,3,5,7,8,10}
count= set()

for i in range(N + 1):
    for j in Numbers:
        if i not in Numbers:
            count.add(i)
print(count)
```

```python
# 6. Write a program to create a new set containing only the even numbers from a given set. 
set1 = { 2, 65, 4, 5, 6, 2, 8, 9, 43, 43, 23, 2}
set2 =set()
for i in set1:
    if i % 2 == 0:
        set2.add(i)
print("even set = ", set2)
```

```python
# 7. Write a program to merge three sets into a single set containing only unique elements. 
set1 = { 2, 65, 4, 5, 6, 2, 8, 9, 43, 43, 23, 2}
set2 = { 9, 45, 6, 2, 8, 78, 34, 56 }
set3 = { 64, 34, 87, 65, 3, 23, 2}
newset = set()
for i in set1:
    if i not in newset:
        newset.add(i)
for j in set2:
    if j not in newset:
        newset.add(j)

for k in set3:
    if k not in newset:
        newset.add(k)

print(newset)        
```
