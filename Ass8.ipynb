```python
# 1. Write a program to count the frequency of each character in a string using a dictionary.
#  - - -  - - Input: "programming" Output: {'p':1, 'r':2, 'o':1, ...} 
str1 = "programming"
dict1 = {}
for ch in str1:
    if ch in dict1:
        dict1[ch] += 1
    else:
        dict1[ch] = 1

print(dict1)            
```

```python
# 3. Write a program to find the key having the maximum value in a dictionary.
#  - - -  - - {'A':85, 'B':92, 'C':78} Output: B 
dict1 = {'A':85, 'B':92, 'C':78}
large = 0
large_key = None

for i in dict1:
    if dict1[i]  > large:
        
        large = dict1[i]
        large_key = i
print(large_key)        
```

```python
# 4. Write a program to find the key having the minimum value in a dictionary. 
dict1 = {'A':85, 'B':92, 'C':78}
min_key = None
min_value = float('inf')

for i in dict1:
    if dict1[i] < min_value:
        min_key = i
        min_value = dict1[i] 
print(min_key, " = " , min_value)        
```
