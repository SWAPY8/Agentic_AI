```python
# 1. Write a program to count the frequency of each character in a string using a dictionary.
#  - - -  - - Input: "programming" Output: {'p':1, 'r':2, 'o':1, ...} 
str1 = "programming"
dict1 = {}
for i in str1:
    if i in dict1:
        dict1[i] += 1
    else:
        dict1[i] = 1

print(dict1)            
```

```python
# 4. Write a program to find the key having the minimum value in a dictionary. 
dict1 = {'A':85, 'B':92, 'C':78}
min_key = None
min_value = float('inf')

for i in dict1:
    if dict1[i] < min_value:
        min_key = i
        min_value = dict1[i] 
print(min_key, " = " , min_value)        
```

```python
# 6. Write a program to remove duplicate values from a dictionary. 
# - - -  - - {'a':10, 'b':20, 'c':10, 'd':30} Output: {'a':10, 'b':20, 'd':30} 
dict1 = {'a':10, 'b':20, 'c':10, 'd':30}
seen = []
result ={}

for key, value in dict1.items():
    if value not in seen:
        seen.append(value)
        result[key] = value
print(result)        
```

```python
# 8. Write a program to find the sum of all values in a dictionary. - {'Math':85, 'Science':90, 'English':80} -  - - Output: 255
dict1 = {'Math':85, 'Science':90, 'English':80}
sum1 = []
for value in dict1.values():
    sum1.append(value)
print(sum(sum1))
```

```python
# 9. Write a program to create a dictionary from two lists: one containing keys and the other containing values. - 
# - Keys = ['A', 'B', 'C'] Values = [10, 20, 30]
Keys = ['A', 'B', 'C'] 
Values = [10, 20, 30]
dict1 = {}
for i in range(len(Keys)):
    dict1[Keys[i]] = Values[i]
print(dict1)     
         
```

```python
# 10. Write a program to create a student result dictionary containing name and marks of 5 students, then display the topper. 
stud_name = ["tatya", "gagu", "anna", "sam", "kenjava"]
stud_marks = [-786, 88, 69, 786, 1 ]
dict1 = {}
for i in range(len(stud_name)):
    dict1[stud_name[i]] = stud_marks[i]
print(dict1)
topper = max(dict1.values())
for key, value in dict1.items():
    if value == topper:
        print("Topper = ", key)
        print("Marks = ", value)
        
```
