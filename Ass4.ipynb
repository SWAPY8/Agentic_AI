```python
# inverted pyramid
rows = 9

for i in range(rows+1):

    #for space
    for j in range(i):
        print(" ", end="")

        # print stars
    for k in range(2 * (rows - i)  + 1) :
        print("*", end="")
    print()       
```

```python
# pyramid

rows = 7
space = 6
starcount = 1

for i in range(rows):
    print(" " * space, "*" * starcount)
    space -= 1
    starcount += 2
```

```python
# 1. Write a program to count the number of vowels and consonants in a given string.
str = "happy holi"
wovel_count = 0
consonents_count = 0

for i in str:
    if i == "a" or i == "e" or i == "i" or i == "o" or i == "u":
        wovel_count += 1
    else:
        consonents_count += 1    

print("wovel_count - ", wovel_count)   
print("consonents_count - ", consonents_count)      
```

```python
# 2. Write a program to check whether a given string is a Palindrome. - - Input: madam Output : Palindrome 
str1 = "swapnil"
is_palidrome = True

for i in range(len(str1)//2):
    if(str1[i] != str1[len(str1) - 1 - i]):
        is_palidrome = False
        break
print("PALINDROME" if is_palidrome  else " not palindrome" )    
```

```python
# 3. Write a program to reverse a string without using slicing ([::-1]) or built-in functions. 

str1 = "success"
temp = ""
i = len(str1) - 1
while i >= 0:
    temp += str1[i]
    i -= 1

print(temp)
```

```python
# 4. Write a program to count the occurrence of each character in a string. 
str1 = "sahilbangar"

visited = ""

for ch in str1:
    if ch not in visited:
        count = 0

        for c in str1:
            if ch == c:
                count += 1

        print(ch, ":", count)

        visited += ch
```

```python

# 5. Write a program to remove all duplicate characters from a string while maintaining their original order. - - Input : programming Output : progamin

str1 = "programming"

visited = ""

for ch in str1:
    if ch not in visited:
        count = 0

        for c in str1:
            if ch == c:
                count += 1

        print(ch, end="")

        visited += ch
```

```python
# Write a program to find the longest word in a sentence. - Input: Python is an amazing programming language  - Output: programming 

str1 = "Python is an amazing programming language"

words = str1.split()
longest = words[0]
for word in words:
    if len(word) > len(longest):
        longest = word
print(longest)         
```

```python
# 7. Write a program to capitalize the first letter of every word without using the title() method. - - Input: 
#    welcome to python Output:   Welcome To Python 
sentence = "welcome to python"
# titleacse = sentence.title()
# print(titleacse)
str1 = ""
words = sentence.split()
# print(words)
for word in words:
    res = word[0].upper() + word[1:]+ " "
    str1 += res
print(str1)
```
