```python
# 1. Write a program using a lambda function to find the square of each number in a list using map(). 
list1 = [8, 3, 5, 1, 9, 7]
square = list(map(lambda x: x*x, list1))
print(square)
```

```python
# 2. Write a program using a lambda function to convert all names in a list to uppercase using map().  
list1 = ["dada", "tatya", "tai", "deepali", "gagan", "sweta"]
upperstring = list(map(lambda x: x.upper(), list1))
print(upperstring)
```

```python
# 3. Write a program using map() to calculate a 10% discount on a list of product prices.  
prices = [600, 234, 645, 873, 876, 222, 876]
disprice = list(map(lambda i: i - (0.1 * i), prices))
print(disprice)
```

```python
# 4. Write a program using filter() to extract all even numbers from a given list.  num
num = [600, 234, 645, 873, 876, 222, 876]
filvalues = list(filter(lambda i: i%2 == 0, num))
print(filvalues)
```

```python
# 5. Write a program using filter() to display all students who scored 35 or more marks. 
student_scores = [44, 23, 45,87, 65,99,32]
goodstudents = list(filter(lambda i: i > 35, student_scores))
print(goodstudents)
```

```python
# 6. Write a program using filter() to remove all empty strings from a list.
str_list = ["hghg", "fstdwy", "","hghg", "fstdwy", "", "rtyhdh"]
# find_empty_str = list(filter(lambda i: i != "" and i != " ", str_list))
find_empty_str = list(filter(lambda i: i.strip(), str_list))
print(find_empty_str)
```

```python
# 7. Write a program using map() to find the length of each string in a list.  
str_list = ["hghg", "fstdwy", "","hghg", "fstdwy", "", "rtyhdh"]
find_len = list(map(lambda i: len(i),str_list))
print(find_len)
```

```python
# 8. Write a program using reduce() to calculate the sum of all numbers in a list. 
from functools import reduce

num_list =[ 78,234, 554,8,324]
find_sum = reduce(lambda i,j: i+j, num_list)
print("total =", find_sum)
```

```python
# 9. Given a list of employee salaries, use map() to increase each salary by 15%.  
emp_sal = [67834, 92343, 42355, 68788, 76543]
inc_sal = list(map(lambda i: i + (0.15*i), emp_sal))
print(inc_sal)
```

```python
# 10. Given a list of integers, use filter() to display only the numbers that are divisible by both 3 and 5.  
nums = [67834, 92343, 42355, 15, 76543]
div_3_5 = list(filter(lambda i: i%3 == 0 and i%5 == 0, nums))
print(div_3_5)
```
