```python
# Type Conversion Program

# int → float
num_int = 25
num_float = float(num_int)
print("int → float:", num_float, type(num_float))

# float → string
price = 99.99
price_str = str(price)
print("float → string:", price_str, type(price_str))

# string → int
number_str = "100"
number_int = int(number_str)
print("string → int:", number_int, type(number_int))

# int → bool
value = 7
value_bool = bool(value)
print("int → bool:", value_bool, type(value_bool))

# bool → int
flag = True
flag_int = int(flag)
print("bool → int:", flag_int, type(flag_int))
```

```python
# 10. Create a program that stores product information. Store: • Product Name  • Price  • Quantity  Calculate: • Total Cost  • GST (18%)  • Final Amount 
product_name = input("Enter Product Name = ")
quantity = int(input("Enter quantity = "))
price = int(input("Enter Product price = "))

total_price = quantity * price
gst = 0.18 * total_price
amt_to_pay = total_price + gst

print("\n----- Product Bill -----")
print("Product Name :", product_name)
print("Price        :", price)
print("Quantity     :", quantity)
print("Total Cost   :", total_price)
print("GST (18%)    :", gst)
print("Final Amount :", amt_to_pay)
```
