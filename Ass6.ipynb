```python
# 1. Write a program to find the largest and smallest element in a tuple without using max() and min(). 
tup1 = (54, 876, 333, 887, 33, 39)
temp = sorted(tup1)
print("min = ", temp[0])
print("max = ", temp[-1])

```

```python
# 2. Write a program to count the number of even and odd numbers in a tuple. 
tup1 =(54, 876, 333, 887, 33, 39)
even = 0
odd = 0

for i in tup1:
    if i % 2 == 0:
        even += 1
    else:
        odd += 1
print("even = ", even)
print("odd = ", odd)            
```

```python
# 3. Write a program to find all common elements between two tuples. 
tup1 =(54, 876, 333, 887, 33, 39)
tup2 =(39, 876, 857, 887, 78, 35)

temp = []

for i in tup1:
    for j in tup2:
        if j == i:
            temp.append(i)
            
temp2 = tuple(temp)
print(temp2)
```

```python
# 4. Write a program to create a new tuple containing only positive numbers from a given tuple. 
tup1 =(54, 8, -76, 333, -887, 33, -39)
temp = []

for i in tup1:
    if i > 0:
        temp.append(i)

tup2 = tuple(temp)
print("Tuple of positive numbers = ", tup2)
```

```python
# 5. Write a program to remove duplicate elements from a tuple while preserving the original order. 
tup1 =(54, 8, 33, -76, 333, 39, -887, 33, -39, 6, -76)
visited = []

for i in tup1:
   if i not in visited:
      visited.append(i)

tup2 = tuple(visited)
print("Tuple of without duplicate numbers = ", tup2)
```

```python
# 6. Write a program to separate all even and odd elements of a tuple into two different tuples. 
# - - -  - - - Input: (10,15,20,25,30) Output: Even = (10,20,30) Odd = (15,25) 
tup1 = (10,15,20,25,30)
even = []
odd = []

for i in tup1:
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)
print("Even Tuple - ", tuple(even))    
print("Odd Tuple - ", tuple(odd))    



```

```python
# 7. Write a program to find the intersection of two tuples. 
tup1 =(54, 876, 333, 887, 33, 39)
tup2 =(39, 876, 857, 887, 78, 35)

intersection = []
for i in tup1:
    for j in tup2:
        if j == i:
            intersection.append(i)
            
print("intersection- ", tuple(intersection))
```

```python
# 8. Write a program to find the union of two tuples (unique elements only). 
tup1 =(54, 876, 333, 887, 33, 39)
tup2 =(39, 876, 857, 887, 78, 35)

union = []

for i in tup1:
    if i not in union:
        union.append(i)

for j in tup2:
    if j not in union:
        union.append(j)

print("Union Values - ", tuple(union))
```

```python
# 9. Write a program to convert a nested tuple like: Input - ((1,2), (3,4), (5,6)) into: output -(1,2,3,4,5,6) 
tup1 = ((1,8), (3,3), (5,9))
temp = []
for i in tup1:
    for j in i:
        temp.append(j)
print(tuple(temp))      
```
