Writing Dictionary Data

```python
import csv
from pathlib import Path

csv_path = Path(r"C:\Users\Admin\Desktop\Agentic_AI") / "student_data.csv" # here we can define path using "Path" 
student_data = [
    {"name": "swapnil", "id": 101, "branch":"aids"},
    {"name": "sarthak", "id": 102, "branch":"entc"},
    {"name": "omakr", "id": 103, "branch":"aids"}

]
header = ["name", "id", "branch"]
with open(csv_path, "w", encoding= "utf-8", newline = "") as file:
    writer = csv.DictWriter(file, fieldnames = header )
    
    writer.writeheader()
    writer.writerows(student_data)
```
