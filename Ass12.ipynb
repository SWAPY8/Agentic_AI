```python
# 2. What will be the output? Explain the LEGB rule step by step. 
x = "Global" 
def outer():     
    x = "Enclosing"      
    def inner():         
        print(x)      
    inner()  
outer() 
# o/p =  Enclosing 
```

```python
# 3. Identify the Local, Enclosing, Global, and Built-in names in the following program: 
x = 10                                  # Global
def outer():     
    y = 20                              # Enclosing
    def inner():        
        z = 30                          # Local
        print(x, y, z, len("Python"))   # In-Built   
    inner() 
outer() 
```

```python
# 4. What will be the output? Is x being shadowed?  
x = 100  
def outer():     
    x = 200      
    def inner():         
        x = 300         
        print(x)      
    inner()   
    print(x) 
outer()
print(x) 

# o/p = 300  200 100 - Yes, x is being shadowed
```

```python
#6. Correct the following program using the global keyword so that the final output is 20.  
x = 10  
def update(): 
    global x
    x = 20  
update() 
print(x) 
 
```

```python
# 7. What will be the output? Explain the difference between the Local and Enclosing 
# Namespace. 
def outer():
    x = 10
    def inner():
        x = 20
        print(x)
    inner()
    print(x)
outer()
# o/p = 20 10. local have limited scope of specific function, and enclosing have larger scope upto the outer functions
```

```python
# 8. What will be the output? Why does nonlocal change the value of x?  
def outer():     
    x = 10      
    def inner():         
        nonlocal x         
        x = 50      
    inner()     
    print(x)  
outer()

#  o/p =  50 . Non local allow to modify variable that has already declared outside that function.
```
