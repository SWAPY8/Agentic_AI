```python
# 1. The Child Class (Can exist on its own)
class Player:
    def __init__(self, name, position):
        self.name = name
        self.position = position
        
    def __str__(self):
        return f"{self.name} ({self.position})"


# 2. The Parent Class (Aggregates the child class)
class Team:
    def __init__(self, team_name):
        self.team_name = team_name
        self.roster = [] # This list will hold Player objects
        
    def add_player(self, player):
        self.roster.append(player)
        
    def show_team(self):
        print(f"--- {self.team_name} Roster ---")
        for player in self.roster:
            print(player)


# --- LET'S TEST IT ---

# Step 1: Create the players INDEPENDENTLY of the team
p1 = Player("Lionel Messi", "Forward")
p2 = Player("Emiliano Martínez", "Goalkeeper")

# Step 2: Create the team
argentina = Team("Argentina National Team")

# Step 3: Aggregate them (Pass the players into the team)
argentina.add_player(p1)
argentina.add_player(p2)

argentina.show_team()

# Step 4: Prove the independent lifespan
del argentina  # The team is completely deleted from memory!

print("\nThe team is gone, but the player still exists:")
print(p1)  # Messi still exists perfectly fine!
```

```python

# 11. Library Management — Aggregation + Encapsulation Create: Book Library • Book should contain title, author, and price.  • Make price private.  • Create getter/setter for price.  • Library should contain multiple Book objects.  • Add books to the library.  • Display all books.  • Search for a book by title.  

class Book:
    def __init__(self, title, author,price):
        self.__price = price
        self.title = title
        self.author = author
        
    def get_price(self):
        return self.__price
    
    def set_price(self, amt):
        if amt > 0:
            self.__price = amt
        else:
            print("Enter corect amt")
            
    def displaybook(self):
        print(f"Book Title = {self.title}, its Auther = {self.author} and its price is {self.__price} ")
        
class Library:
    def __init__(self):
        self.books = []
       
    def add_book(self, book_obj):
        self.books.append(book_obj)
        
    def display(self):
        for book in self.books:
            book.displaybook()
            
    def search_book(self, ttl):
        for book in self.books:
            if book.title.lower() == ttl.lower():
                book.displaybook()
                return
            
        
        
        
        
    
b1 = Book("Zep","Akash Wagh", 600)
b2 = Book("Agni","Abdul kalm", 1600)
l1 = Library()

l1.add_book(b1)
l1.add_book(b2)

b1.set_price(800)

l1.display()
l1.search_book("zep")
```

```python
# 12. E-Commerce Product — Encapsulation + Static Method Create a Product class. • Private variable __price.  • Instance variables: name, quantity.  • Getter/setter for price.  • Instance method total_price().  • Static method calculate_gst(price) that calculates GST.  • Display the final product price including GST.  

class Product:
    
    gst_perc = 18
    
    def __init__(self,name,quantity):
        self.name = name
        self.quantity = quantity
        self.__price = 800
        
    def get_price(self):
        return self.__price 
    
    def set_price(self, new_amt):
        if new_amt > 0:
            self.__price = new_amt
        else:
            print("amt cannt be negative....")
    
    def total_price(self):
        total = self.__price * self.quantity
        return total
    
    @staticmethod
    def calculate_gst(base_total):
        return base_total * (Product.gst_perc / 100)
         
    
    def total_payable(self):
        base = self.total_price()
        total_with_gst = base + self.calculate_gst(base)
        
        print(f"""
your quantity was : {self.quantity},
price for 1 piece is {self.__price},
Total base price : {base},
Bill : {total_with_gst}
              """)
        
        
        
        
        
p1 = Product("pant" , 90)
p1.total_payable()
```
