```python
# 3. Create an Employee parent class with name and salary. Create a Manager child class with an additional department attribute. Display all details using constructors. 

class Employee:
  def __init__(self, name, salary):
    self.name = name
    self.salary = salary

class Manager(Employee):
    def __init__(self, dept_id, name, salary, experience):
        super().__init__(name, salary)
        self.dept_id = dept_id
        self.experience = experience
        
    def display(self):
        return f"{self.dept_id} {self.name} {self.salary} {self.experience}"

m1 = Manager(101, "Sarthak", 99, 4.5)
m1.display()
```

```python
# 5. Create a Person parent class with a display() method. Create a Student child class that overrides display() but also calls the parent display() using super().

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
    def display(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")

class Student(Person):
    def __init__(self, name, age, student_id):
        super().__init__(name, age)
        self.student_id = student_id
        
    def display(self):
        super().display()
        
        print(f"Student ID: {self.student_id}")

student1 = Student("Aisha", 21, "S-2048")
student1.display()
```

```python

# 7. Bank Account Create a BankAccount class with a private attribute __balance.  Create get_balance() to view the balance.  Create set_balance() to update the balance only if the amount is valid.  Create a withdraw() method with proper balance validation
class BankAccount:
    def __init__(self, balance = 0):
        
        self.__balance = balance
        
    # getter
    def get_balance(self):
        return self.__balance
    
    # setter
    def set_balance(self , amount):
            
            if amount >= 0:
                self.__balance = amount
            else:
                print("enter valid amt.")
            return self.__balance
    
class Operation(BankAccount):
    def __init__(self, balance = 0):
        super().__init__(balance)
        self.initial_bal = balance
        
    # withdraw    
    def withdraw(self, withdraw_amt):
        current = self.get_balance()
        if current <= 0:
            print("zero balance.")
        elif withdraw_amt > current:
            print("Insufficient amt.")
        else:
            self.set_balance(current - withdraw_amt)  
            # print(current)   
        return current - withdraw_amt      
     
b1 = Operation(7000)
print(f" Current balance is {b1.get_balance()}")

print(f" Current balance after set bal. - {b1.set_balance(90000)}")

print(f" Current balance after withdrawal- {b1.withdraw(700)}")
```

```python
# 8. Student Marks Create a Student class with private attributes __name and __marks.  Create getter and setter methods for marks.  Allow marks only between 0 and 100.  Create a child class ExamStudent that inherits from Student 

class Student:
    def __init__(self, name, mark):
        self.__name = name
        self.__mark = mark
    
    def get_mark(self):
        return self.__mark
    
    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__mark = marks
        else:
            print("enter valid marks.")
        return self.__mark
    
class ExamStudent(Student):
    def __init__(self, name, mark):
        super().__init__(name, mark)
        
e1 = ExamStudent("swapnil", 90)
e1.get_mark()
e1.set_marks(99)
```
